```python
import google_colab_selenium as gs
import time
from IPython.display import Image, display

# 1. Initialize the Chrome driver (automatically handles headless configuration)
driver = gs.Chrome()

# 2. Navigate to the website
url = "https://fast.com/"

driver.get(url)
time.sleep(60)
# 3. Take and save the screenshot
screenshot_path = "screenshot.png"
driver.save_screenshot(screenshot_path)

# 4. Display the screenshot in the Colab output
display(Image(screenshot_path))

# 5. Close the driver
driver.quit()
```
